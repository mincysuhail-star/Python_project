 Data Preprocessing
</h2>

This step ensures the dataset is clean,consistent,and suitable for analysis.The cleaning steps include (i)**Correcting Data Types**,(ii)**Removing duplicates**,(iii)**Creating Derived columns**,(iv)**Standardizing column names and Categorical values**(v)****Handling missing values****(vi)**Removing unnecessary columns**,(vii)**Filtering data**.


### (1)Correcting column name into a unique format
Changing all column names into lower and replaced white spaces with _

```python
#Standardizing Work Type Categories
df.columns=(
    df.columns.
    str.strip()
    .str.lower()
    .str.replace(" ","_")
)
```

```python
df.columns
```

```python
df.dtypes
```

### (2)Correcting Data Types
In this step,the data types of certain columns are corrected to ensure consistent and efficient analysis.This include fixing the **customer satisfaction rate ** to interger format column and updating **categorical** columns ,**Time of resolution, First response time , date of purchase** are corrected into date format ,and **gender** as categoricatl format

```python
df["customer_satisfaction_rating"]=pd.to_numeric(df["customer_satisfaction_rating"].astype('Int64'))
```

```python
df["time_to_resolution"]=pd.to_datetime(df["time_to_resolution"],errors="coerce")
```

```python
df["first_response_time"]=pd.to_datetime(df["first_response_time"],errors="coerce")
```

```python
df["date_of_purchase"]=pd.to_datetime(df["date_of_purchase"],errors="coerce")
```

```python
df.dtypes
```

### (3)Removing Duplicates
First, checking the number of rows which are duplicated and removing the duplicates 

```python
#Checking the number of duplicate rows
print("Duplicate Rows Found:",df.duplicated().sum())
```

```python
#Removing duplicates
print("Number of Duplicate Rows:",df.drop_duplicates(inplace=True))
```

***Info:-*** The dataset was checked for duplicate entries using df.duplicated().sum().The result returned 0 duplicates,indicating that no repeated rows were found.The drop_duplicates() function was applied as a standard preprocessing step,although no removal was required.

### (4)Categorical Column
1)Making gender of the customer column as categorical format of "male','female',and other 
2)Making ticket type column into 4 catgories 

```python
##Creating gender Category
gender_category = pd.CategoricalDtype(
    categories=["Male", "Female", "Other"],
    ordered=False
)

df["customer_gender"] = df["customer_gender"].astype(gender_category)
```

```python
#Converting tikcet type  To Category
tikcet_category=["Billing inquiry","Technical issue","Cancellation request","Product inquiry","Refund request"]
df["ticket_type"]=df["ticket_type"].astype("category")
```

### (5)Handling Missing Values
The dataset contained **333** missing(NaN)values in the **resolution** column,**474** values in **first_response_time**,**5** in **customer_satisfaction_rating** and in.Since bmi is a numerical feature,these missing/Nan entries were replaced using the median,which is more robust to outliers and preserves the overall distribution.This ensures the dataset remains complete and suitable for further preprocessing.

```python
df.isnull().sum()
```

```python
df["resolution"].nunique()
```

```python
df["first_response_time"].nunique()
```

```python
df["customer_satisfaction_rating"].nunique()
```

```python
df["time_to_resolution"].nunique()
```

```python
df["customer_satisfaction_rating"]=df["customer_satisfaction_rating"].fillna(df["customer_satisfaction_rating"].median())
df["time_to_resolution"]=df["time_to_resolution"].fillna(df["time_to_resolution"].ffill().bfill())
df["first_response_time"]=df["first_response_time"].fillna(df["first_response_time"].ffill().bfill())
df["resolution"] = df["resolution"].fillna(df["resolution"].mode()[0])
```

```python
df.isnull().sum()
```

### (6)Creating Derived Columns

```python
#(1)Creating Age Group column
df["age_category"]=np.where(df["customer_age"]<=21,"Adult",np.where(df["customer_age"]<=45,"Middle Aged","Aged"))
df["age_category"]
```

***Info:-*** Created an **Age Group** categorical feature using where condition from 21 ols customer coming into the adult category, upto 45 as middle aged and remaining as aged

```python
#(2)creating ticket_priority_level
df["ticket_priority"].unique()
df["ticket_priority_level"]=np.where(df["ticket_priority"]=="Critical", "P1",np.where(df["ticket_priority"]=="Low", "P16",np.where(df["ticket_priority"]=="High","P4","P8")))
df["ticket_priority_level"]
```

***Info:-***  Categorized the ticket priorities to different priority levels, for critical priority  used P1 level ,for high P4 , for medium P8 and P16 for low priority tickets.

```python
df.dtypes
```

```python
##(3)Creating Time taken
df["resolution_time"] = (
    df["time_to_resolution"] - df["first_response_time"]
)
df["resolution_time"]

df["resolution_time"].unique().tolist()
```

##(4)Ticket Type Category
To categorise the ticket type using the ticket type category 
for Billing related categorised to Account
Technical issue  to IT 
Refund request issue Payment 
Cancellation request to Cancelled 
Product Inquiry to Product related 


```python
## Creating ticket_type_category
df["ticket_type_category"]=np.where(df["ticket_type"]=="Technical issue","IT",
                                    np.where(df["ticket_type"]=="Billing inquiry","Account",
                                             np.where(df["ticket_type"]=="Refund request","Payment",
                                                      np.where(df["ticket_type"]=="Cancellation request",
                                                               "Cancelled","Product "))))
df["ticket_type_category"]
```
